## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
def convert_case(text):

    if "_" in text or text.islower():
        mode = "to_pascal"
    else:
        mode = "to_snake"

    if mode == "to_pascal":
        if "_" not in text and text[0].isupper():
            return text

        words = text.split("_")
        result = ""

        for word in words:
                result += word.capitalize()

        return result

    if mode == "to_snake":
        if text.islower():
            return text

        result = ""

        for i in range(len(text)):
            char = text[i]

            if char.isupper() and i > 0:
                prev_char = text[i - 1]

                if prev_char != "_" and not prev_char.isupper():
                    result += "_"

            result += char.lower()

        return result
```

```python
convert_case("otus_course")
```

```python
convert_case("PythonIsTheBest")
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
def is_valid_date(date_str):

    # Разбиваем строку
    parts = date_str.split(".")
    if len(parts) != 3:
        return False

    day_str = parts[0]
    month_str = parts[1]
    year_str = parts[2]

    # Проверяем, что только цифры
    if not (day_str.isdigit() and month_str.isdigit() and year_str.isdigit()):
        return False

    day = int(day_str)
    month = int(month_str)
    year = int(year_str)

    # Проверка для года и месяца
    if year < 1:
        return False
    if month < 1 or month > 12:
        return False

    # Високосный год
    # Год високосный, если он делится на 400 ИЛИ (делится на 4 И не делится на 100)
    is_leap = False
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        is_leap = True

    # Максимальное количество дней в месяце
    if month == 2:
        if is_leap:
            max_days = 29
        else:
            max_days = 28
    elif month in (4, 6, 9, 11):
        max_days = 30
    else:
        max_days = 31

    if day >= 1 and day <= max_days:
        return True
    else:
        return False
```

```python
is_valid_date('29.02.2000')
```

```python
is_valid_date('29.02.2001')
```

```python
is_valid_date('31.04.1962')
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
def is_simple(number):

    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True
```

```python
print(is_simple(17))
print(is_simple(20))
print(is_simple(23))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
# Проверка имени и фамилии
def check_name(text):
    if not text.isalpha():
        return None
    return text.capitalize()

# Проверка возраста
def check_age(text):
    if not text.isdigit():
        return None
    age = int(text)

    if 18 <= age <= 60:
        return age
    return None


# Проверка ID
def check_id(text, existing_users):
    if not text.isdigit():
        return None

    if len(text) > 8:
        return None

    formatted_id = text.zfill(8)

    if formatted_id in existing_users:
        print("Ошибка: Пользователь с таким ID уже существует!")
        return None

    return formatted_id
```

```python
def print_users(users_dict):
    if len(users_dict) == 0:
        print("Список пустой")
        return

    for uid, data in users_dict.items():
        name, surname, age = data
        print(f"Имя: {name}, Фамилия: {surname}, Возраст: {age}, ID: {uid}")
```

```python
users = {}

print("Введите данные пользователей. Чтобы завершить ввод, оставьте любое поле пустым и нажмите Enter")

while True:
    # 1. Ввод имени
    in_name = input("Введите имя: ")
    if in_name == "":
        break

    name = check_name(in_name)
    if name is None:
        print("Ошибка: Имя должно состоять только из букв")
        continue

    # 2. Ввод фамилии
    in_surname = input("Введите фамилию: ")
    if in_surname == "":
        break

    surname = check_name(in_surname)
    if surname is None:
        print("Ошибка: Фамилия должна состоять только из букв")
        continue

    # 3. Ввод возраста
    in_age = input("Введите возраст (от 18 до 60): ")
    if in_age == "":
        break

    age = check_age(in_age)
    if age is None:
        print("Ошибка: Возраст должен быть числом от 18 до 60")
        continue

    # 4. Ввод ID
    in_id = input("Введите ID (целое число, до 8 цифр): ")
    if in_id == "":
        break

    user_id = check_id(in_id, users)
    if user_id is None:
        print("Ошибка: Некорректный ID")
        continue

    users[user_id] = (name, surname, age)
```

```python
print_users(users)
```
